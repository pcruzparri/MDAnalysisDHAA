```python
import MDAnalysis as mda
import matplotlib.pyplot as plt
import os
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
script_dir = os.getcwd()

pdb_path = os.path.join(script_dir, '4Zn-8Ser pH', 'step4_pbcsetup.pdb')
dcd_path = os.path.join(script_dir, "Traj 4Zn-8Ser pH 1fs.dcd")

u = mda.Universe(pdb_path, dcd_path)
```

```python
set(atom.name for atom in u.atoms)
```

```python
zn_atoms = u.select_atoms("name ZN")
c_alpha = u.select_atoms('name C3')
```

```python
[np.linalg.norm(zn.position - c_alpha[0].position) for zn in zn_atoms]
```

```python
c_carboxyl = u.select_atoms('name C2')
distances = []
for atom in u.atoms:
    distance = np.linalg.norm(atom.position - c_carboxyl.positions[0])
    distances.append((atom.name, distance))


print(sorted(distances, key=lambda x: x[1]))
```
